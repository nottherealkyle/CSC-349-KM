1. What challenges did you encounter when loading and inspecting a raw external dataset from Kaggle compared to pre-packaged datasets, specifically regarding implicit data types like whitespace entries in TotalCharges?

    The challenges I faced when loading and inspecting a raw external dataset were formatting errors such as whitespace entries in TotalCharges. In that column, what should have been numeric were read as objects, as customers with tenure == 0 and as a result were not yet billed, so a blank space was placed as the value rather than 0. In a pre-packaged dataset, this would not be a problem and it would have been classified properly as numeric.
2. How did you determine which columns required missing value imputation versus other data cleaning methods, and what strategy did you choose?

    I determined which columns required missing value imputation by running df.isna().sum(), which showed me that only that TotalCharges was missing data so that is where I focused the missing value imputation. I chose median as my strategy because there were a few large outliers that skewed the mean, therefore I believe that median was the best option.
3. Explain how the Scikit-Learn ColumnTransformer and Pipeline streamline the data preparation process and prevent data leakage across preprocessing and training steps.
    
    Without ColumnTranformer, the numeric and categorical columns would each need their own manually written imputation, scaling, and encoding code, all of which applied seperately before being handed to the classifier. Alongside this, putting everything into a Pipeline lets the whole process run with one .fit() call. It also prevents data leakage because the Pipeline only fits on the training data, so the imputation, scaling, and encoding values all come from X_train and are just reused on X_test instead of being recalculated.
4. How did comparing StandardScaler versus MinMaxScaler impact your model training, convergence, or final evaluation metrics?

    I only used StandardScaler, so I didn't directly compare it to MinMaxScaler. I think StandardScaler was the better fit anyway since TotalCharges and MonthlyCharges have some outliers, and MinMaxScaler would have squished most customers into a small range because it scales off the min and max.

5. How does configuring SGDClassifier(loss='log_loss') differ from standard linear classification models, and what advantages do probability outputs and ROC-AUC metrics provide when evaluating customer churn?

    By default SGDClassifier uses hinge loss and only outputs a class label. Setting loss='log_loss' makes it act like logistic regression instead, so predict_proba() gives an actual probability. That's useful for churn since the classes are imbalanced (about 73% No vs 27% Yes), so accuracy alone can be misleading. Probabilities let the company rank customers by risk instead of just getting a yes/no, and ROC-AUC gives a fairer measure of performance than accuracy since it looks across all thresholds instead of just one.
6. Based on your model coefficients or feature evaluation, which features show the strongest relationship with customer churn, and what operational business insights do these findings reveal?

    InternetService_Fiber optic and Contract_Month-to-month had the strongest positive coefficients, so they pushed predictions toward churn the most, while tenure had the strongest negative coefficient, pushing toward retention. This suggests the company could focus on moving customers off month-to-month contracts, maybe with a discount for switching to a longer term. The fiber optic result was a bit surprising since it's the better internet option, so that's a group worth looking into further. Tenure being the top retention factor also makes sense, since it shows customers are most likely to leave earlier on, so the first few months of the relationship probably matter most for reducing churn.
