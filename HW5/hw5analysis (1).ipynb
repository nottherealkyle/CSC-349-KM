Explain how implicit missing values impact medical datasets, and justify your choice of imputation strategy when handling these zeros compared to outright row deletion.

  1. Implicit missing values are a problem because the computer treats them as real measurements. This file had no empty cells, but there were 5 glucose, 35 blood pressure, 227 skin thickness, 374 insulin and 11 BMI readings of 0, which is impossible for a living patient. They pull the averages down (mean insulin was 79.8 with the zeros and 155.5 without) and make the scaler, SVC and Logistic Regression treat 0 as a real, extreme patient. I filled them with the median because insulin and skin thickness are skewed, and I did not delete rows because 376 of the 768 patients have at least one missing value, which would leave only 392 patients.


Contrast how hard voting aggregates predictions versus soft voting. Why do well-calibrated probability estimates matter when implementing soft voting across heterogeneous models?

  2. Hard voting lets each model cast one vote and the majority class wins, so a model that is 51% sure counts the same as one that is 99% sure. Soft voting averages the predicted probabilities and picks the class with the highest average, so confident models have more influence. This only works if the probabilities mean the same thing for every model, and Logistic Regression, the SVC and a Decision Tree produce them in very different ways, so an overconfident model can dominate the average without being more accurate. Both of my ensembles scored 72.7%, below the best single model (Decision Tree, 76.0%). Hard voting found 30 of the 54 diabetic patients and soft voting found 27, so soft voting was a little more cautious about predicting diabetes.


What is the role of out-of-bag evaluation in Bagging classifiers, and why does an OOB score serve as an efficient alternative to traditional cross-validation splits during ensemble training?

  3. Each tree in Bagging is trained on a bootstrap sample, which leaves out about 37% of the training patients. The OOB score predicts each patient using only the trees that never saw that patient, so it is an honest estimate of performance on new data. It is efficient because it comes from the single training run, while 5-fold cross-validation would train all 200 trees 5 times and would need a validation set taken out of the 614 training patients. My OOB score was 75.9% and the test accuracy was 74.0%, so it was a good stand-in for the test score.


How does introducing random feature subsetting in Random Forests differ from standard Bagging, and how does this feature diversity impact model variance and generalization?

  4. Bagging only changes which patients each tree sees, so every split can use all 8 features and the trees tend to pick the same strong features (glucose and BMI). Random Forest also limits each split to a random subset of features (about 3 of 8), so the trees look different from each other, and averaging trees that make different mistakes reduces variance more. The forest had a higher OOB score than Bagging (77.0% vs 75.9%) but a slightly lower test accuracy (73.4% vs 74.0%), a difference of one patient out of 154, so I cannot say either is really better here. The forest's importances showed glucose as the strongest feature (0.268), followed by BMI (0.160).


Discuss the mechanics of sequential sample weighting in AdaBoost. Based on your training versus test accuracy results, did AdaBoost exhibit signs of overfitting, and how can hyperparameter adjustments mitigate this?

  5. AdaBoost starts with equal weights on every training patient and trains a decision stump. It then raises the weights of the patients that stump got wrong so the next stump focuses on them, and each stump gets a larger vote if it is more accurate. My AdaBoost had 82.3% training accuracy and 76.0% test accuracy, and its best test accuracy (76.6%) came after only 21 stumps, so I think it overfit mildly. Fewer stumps or a lower learning_rate should reduce this, and in a quick check on the same split (not in the notebook) a learning rate of 0.1 shrank the gap to about 3 points, though test accuracy did not improve. I would pick these settings with GridSearchCV on the training set.
