How do the feature types in the UCI Heart Disease dataset (combining continuous physiological telemetry like chol and thalach with categorical/ordinal indicators like cp and thal) differ from previous datasets, and why is proper scaling critical for Support Vector Machines compared to Decision Trees?

  1. The heart disease dataset mixes numeric columns like chol and thalch with text categories like cp and thal, so I had to turn the categories into 0/1 columns, which took 13 columns up to 28. Scaling matters for the SVC because it measures how far apart patients are, and a column like chol (up to 603) would overpower oldpeak (up to about 6) just because its numbers are bigger. A Decision Tree only asks about one column at a time, so the size of the numbers in other columns does not affect it.


Explain how hidden missing values (such as ? symbols) impact raw medical datasets, and justify your choice of imputation strategies for skewed numerical vitals versus categorical diagnostic fields.

  2. Hidden missing values are a problem because the computer treats them as real data. This file had no ? symbols, but 172 patients had a cholesterol of 0, which is impossible, and it made the average look like 199 instead of about 247. I filled the numeric vitals with the median because a few very high readings pull the mean away from a typical patient, and I filled the categorical fields with the most frequent value because an average of categories does not exist.


Contrast how a Support Vector Classifier (using an RBF kernel) constructs decision boundaries versus how a Decision Tree recursively partitions feature space. What are the interpretability tradeoffs between the two models?

  3. The SVC draws one smooth, curved boundary using all the features together, while the Decision Tree asks yes/no questions about one feature at a time and splits the patients into smaller and smaller groups. The SVC was more accurate in my results (82.6% vs 76.1% test accuracy), but it cannot explain why it made a prediction. The tree is easier to explain because I can plot it and follow any patient from the top to the bottom.


Discuss how tuning hyperparameters such as regularization parameter C and kernel coefficient gamma in an SVC impacts the trade-off between bias, variance, and model generalization on unseen medical test data.

  4. C controls how strictly the model tries to get every training patient right, and gamma controls how closely the boundary follows individual patients. When both were too low (C=0.1, gamma=0.001), the model was too simple and only scored 55.3%, which is just guessing the most common answer (high bias). When both were too high (C=100, gamma=1), validation accuracy fell to 69.8% because the model memorized the training patients (high variance), while the best setting (C=1, gamma="scale") scored 82.5% in validation and 82.6% on the test set.


Inspecting your Decision Tree feature importances or model coefficients, which clinical indicators (e.g., maximum heart rate, chest pain type, ST depression) exhibit the strongest predictive power for heart disease, and what preventative screening insights do these results provide for cardiologists?

  5. Chest pain type was the strongest, since my tree's first question is whether the patient has asymptomatic chest pain, followed by age, oldpeak (ST depression) and thalch (maximum heart rate). About 79% of patients with no chest pain had heart disease compared to about 28% of the others, so cardiologists should not treat a lack of chest pain as a sign that a patient is fine. The tree's use of ST depression and maximum heart rate also suggests exercise stress tests are useful for separating high-risk and low-risk patients.
