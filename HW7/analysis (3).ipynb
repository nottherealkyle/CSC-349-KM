Why is feature scaling essential prior to executing K-Means, DBSCAN, or HDBSCAN on the Wine dataset? Specifically, how would unscaled features distort distance calculations relative to other features? Provide specific examples.

  1. All three algorithms work from distances between wines, so a column with big numbers counts for more than a column with small numbers. Proline has a standard deviation of 314.9 (values from 278 to 1680) while hue has 0.229 and nonflavanoid phenols has 0.124. When I checked how much of the squared distance between two wines each raw feature gives, proline gave 99.77%, magnesium 0.21% and the other 11 features together less than 0.03%, so without scaling the algorithms would really be clustering on proline alone and flavanoids or color intensity would hardly matter. For DBSCAN and HDBSCAN the eps radius and the core distances would also only be a range of proline values. After StandardScaler every feature has a mean of 0 and a standard deviation of 1, so each one gets a fair share of the distance.


How did the Elbow Method and Silhouette Analysis in Task 2 reflect the natural structure of the Wine dataset? Did both evaluation metrics align with the true number of wine cultivars (k=3)? Explain any discrepancies observed.

  2. Both metrics suggested k=3, which is the true number of cultivars. The inertia fell from 2314.00 at k=1 to 1658.76 at k=2 and 1277.93 at k=3, but only to 1175.43 at k=4, and the silhouette score was highest at k=3 (0.2849) compared with 0.2593 at k=2 and 0.2602 at k=4. There was no discrepancy in k, but the structure is not very clean. The elbow is gradual and the best silhouette is low, which means the three cultivars touch each other. This agrees with Task 4, where the two ambiguous wines were only 0.124 and 0.165 closer to their own centroid than to the next one.


Compare the clustering behavior and cluster assignments of K-Means against DBSCAN and HDBSCAN on the 13-dimensional Wine feature space. Why did DBSCAN and HDBSCAN classify specific wine samples as noise, while K-Means assigned every sample to a cluster?

  3. K-Means gave all 178 wines to one of 3 clusters, while DBSCAN (eps=2.25, min_samples=5) found 2 clusters of 90 and 42 wines with 46 noise points, and HDBSCAN found 2 clusters of 86 and 34 wines with 58 noise points. K-Means has no noise because every wine goes to its nearest centroid, so even Wine 121, which was 6.16 to 7.48 away from every centroid, got a cluster. DBSCAN and HDBSCAN only keep a wine in a cluster if enough neighbors are close to it, so wines in sparse places are labeled noise, and in the PCA plots these are mostly in the band between the two groups. Most of the noise was cultivar 1 (33 of 46 for DBSCAN and 37 of 58 for HDBSCAN), which is about half of the 71 cultivar 1 wines, compared with 6 of the 59 cultivar 0 wines. Neither method found 3 clusters because the groups are joined by wines in the middle, so they were merged into one.


Contrast the operational mechanisms of standard DBSCAN and HDBSCAN. How does HDBSCAN's hierarchical, stability-based approach overcome DBSCAN's vulnerability to a single global eps parameter when clustering chemical datasets with varying local densities?

  4. DBSCAN uses one eps and one min_samples for the whole dataset, and it grows clusters by joining core points that are within eps of each other. This was very sensitive to eps in my grid: with min_samples=5, eps=1.5 made all 178 wines noise, eps=2.0 gave 5 clusters but 85 noise points, eps=2.25 gave 2 clusters, and eps=2.5 and above gave only 1 cluster, so no single eps gave the 3 cultivars. HDBSCAN builds the single-linkage tree over all distances (the dendrogram in Task 7, where cutting at eps=2.25 still leaves 35 groups), removes branches smaller than min_cluster_size and keeps the most stable clusters, so a dense cluster and a sparse cluster can each be kept at the distance that suits them. In my results it did not find more clusters than DBSCAN (2 and 2) and it had more noise (58 vs 46) because min_samples=5 is strict, but it needed no search over eps.


Based on your comparative classification experiment in Task 8, how effectively did the cluster assignments from K-Means, DBSCAN, and HDBSCAN enable a downstream classifier to predict the true ground-truth wine cultivars (Class 0-2)? Discuss the limitations of mapping unsupervised clusters to supervised class labels when dealing with noise and density-based groupings.

  5. K-Means labels predicted the cultivars best with 97.2% test accuracy (35 of 36 wines), compared with 80.6% for DBSCAN and 75.0% for HDBSCAN. The silhouette scores went the other way (0.2849, 0.3371 and 0.3524) because DBSCAN and HDBSCAN were only scored on the wines they did not call noise (132 and 120 wines), so a higher score does not mean a better match to the cultivars. With only 2 clusters and a noise group the classifier has just 3 values to work with, and the noise group mixes cultivars (6 of class 0, 33 of class 1 and 7 of class 2 for DBSCAN), so it cannot separate 3 classes well. The cluster numbers are also arbitrary (cluster 0 is not Class 0), and one test wine is worth about 2.8 points. So density-based clusters describe dense regions, not classes, and they should not be judged only by how well they predict labels.
